```python
# Genspark Agentic News Curation System
# Stage 1: Non-Agentic Baseline

PROJECT_TITLE = "Genspark Agentic News Curation System Inspired by Genspark"

user_query = "Artificial Intelligence"

print("Project:", PROJECT_TITLE)
print("User Query:", user_query)


def baseline_answer(user_query):

    if not isinstance(user_query, str):
        return "Invalid user query."

    user_query = user_query.strip()

    if user_query == "":
        return "Please enter a news topic or question."

    return (
        "Baseline received the user query: "
        + user_query
        + "\n"
        + "This baseline does not use external news sources yet."
    )


print()
print("Baseline Answer:")
print(baseline_answer(user_query))
```
