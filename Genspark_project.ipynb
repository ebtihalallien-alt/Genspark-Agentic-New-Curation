```python
import requests
import feedparser
import sqlite3
from getpass import getpass

print("Libraries are ready.")
```

```python
from getpass import getpass

newsdata_api_key = getpass("Enter NewsData.io API Key: ")

print("NewsData.io API key is ready.")
```

```python
user_query = "Artificial Intelligence"

print("User Query:", user_query)
```

```python
newsdata_url = "https://newsdata.io/api/1/latest"

newsdata_response = requests.get(
    newsdata_url,
    params={
        "apikey": newsdata_api_key,
        "q": user_query,
        "language": "en"
    }
)

newsdata_data = newsdata_response.json()

print("NewsData.io status:", newsdata_response.status_code)
print("Number of articles:", len(newsdata_data.get("results", [])))
```

```python
rss_url = "https://news.google.com/rss/search?q=Artificial+Intelligence&hl=en-US&gl=US&ceid=US:en"

rss_feed = feedparser.parse(rss_url)

print("RSS feed loaded.")
print("Number of articles:", len(rss_feed.entries))
```

```python
rss_articles = []

for entry in rss_feed.entries[:10]:
    rss_articles.append({
        "title": entry.get("title", ""),
        "description": entry.get("summary", ""),
        "link": entry.get("link", ""),
        "source": "RSS"
    })

print("RSS articles collected:", len(rss_articles))
print(rss_articles[0])
```

```python
from getpass import getpass

gnews_api_key = getpass("Enter GNews API Key: ")

print("GNews API key is ready.")
```

```python
gnews_url = "https://gnews.io/api/v4/search"

gnews_response = requests.get(
    gnews_url,
    params={
        "q": user_query,
        "lang": "en",
        "max": 10,
        "apikey": gnews_api_key
    }
)

gnews_data = gnews_response.json()

print("GNews status:", gnews_response.status_code)
print("Number of articles:", len(gnews_data.get("articles", [])))
```

```python
gnews_articles = []

for article in gnews_data.get("articles", []):
    gnews_articles.append({
        "title": article.get("title", ""),
        "description": article.get("description", ""),
        "link": article.get("url", ""),
        "source": "GNews"
    })

print("GNews articles collected:", len(gnews_articles))
print(gnews_articles[0])
```

```python
newsdata_articles = []

for article in newsdata_data.get("results", []):
    newsdata_articles.append({
        "title": article.get("title", ""),
        "description": article.get("description", ""),
        "link": article.get("link", ""),
        "source": "NewsData.io"
    })

print("NewsData.io articles collected:", len(newsdata_articles))
```

```python
print("RSS articles ready:", len(rss_articles))
print(rss_articles[0])
```

```python
all_articles = newsdata_articles + rss_articles + gnews_articles

print("Total articles:", len(all_articles))
print("NewsData.io:", len(newsdata_articles))
print("RSS:", len(rss_articles))
print("GNews:", len(gnews_articles))
```
