```python
"""
Genspark Agentic News Curation System
Stage 5: Tool Validation and Policy

Project tools:
1. NewsData.io
2. GNews
3. Google News RSS

Purpose:
The agent can propose a tool, but the application validates
the tool request before execution.
"""

ALLOWED_TOOLS = {
    "newsdata",
    "gnews",
    "google_news_rss"
}


TOOL_CATALOG = {
    "newsdata": {
        "purpose": "Collect recent public news articles.",
        "source": "NewsData.io",
        "side_effect": "None"
    },

    "gnews": {
        "purpose": "Collect recent public news articles.",
        "source": "GNews",
        "side_effect": "None"
    },

    "google_news_rss": {
        "purpose": "Collect recent public news through RSS.",
        "source": "Google News RSS",
        "side_effect": "None"
    }
}


print("Available Tools:")
for tool_name in TOOL_CATALOG:
    print("-", tool_name)
```

```python
def validate_tool_request(tool_name, user_query):

    # Schema validation
    if not isinstance(tool_name, str):
        return False, "Tool name must be text."

    if not isinstance(user_query, str):
        return False, "User query must be text."

    # Reference validation
    if tool_name not in ALLOWED_TOOLS:
        return False, "Tool is not in the approved tool catalog."

    # Input validation
    if user_query.strip() == "":
        return False, "News topic cannot be empty."

    # Policy validation
    forbidden_actions = [
        "purchase",
        "buy",
        "payment",
        "transfer money",
        "financial transaction"
    ]

    query_lower = user_query.lower()

    for action in forbidden_actions:
        if action in query_lower:
            return False, "Request violates the project safety policy."

    return True, "Tool request is approved."


user_query = "Find recent news about Artificial Intelligence."

requested_tool = "newsdata"


is_valid, message = validate_tool_request(
    requested_tool,
    user_query
)


print("Tool Request:")
print(requested_tool)

print()
print("Validation Result:")
print(message)
```
