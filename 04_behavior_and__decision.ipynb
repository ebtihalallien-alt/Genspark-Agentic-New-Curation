```python
"""
Genspark Agentic News Curation System
Stage 4: Behaviour Specification and Structured Decision

Concepts:
    - Behaviour Specification
    - Structured Output
    - Semantic / Policy Validation

Purpose:
    Define what the agent is allowed to do and represent
    its decision in a structured form.
"""


ALLOWED_ACTIONS = {
    "answer",
    "tool",
    "ask_user",
    "escalate"
}


BEHAVIOUR_SPEC = {
    "ROLE": "News curation assistant",
    "GOAL": "Collect and organize relevant public news for the user.",
    "AVAILABLE_INFORMATION": [
        "User query",
        "News gathered from approved public sources"
    ],
    "AVAILABLE_TOOLS": [
        "NewsData.io",
        "GNews",
        "Google News RSS"
    ],
    "RULES": [
        "Use public information only.",
        "Identify the news source.",
        "Do not perform purchases.",
        "Do not perform financial actions.",
        "Do not perform irreversible actions."
    ],
    "OUTPUT_CONTRACT": [
        "Return a structured decision.",
        "Use only an allowed action.",
        "Provide a reason for the decision."
    ]
}


def validate_decision(decision):

    if not isinstance(decision, dict):
        return False, "Decision must be a dictionary."

    required_fields = [
        "action",
        "reason"
    ]

    for field in required_fields:

        if field not in decision:
            return False, "Missing field: " + field

    action = decision["action"]

    if action not in ALLOWED_ACTIONS:
        return False, "Action is not allowed."

    reason = decision["reason"]

    if not isinstance(reason, str) or reason.strip() == "":
        return False, "Reason must be non-empty text."

    return True, "Decision is valid."


def create_decision(user_query):

    if not isinstance(user_query, str):
        return {
            "action": "ask_user",
            "reason": "The user query must be text."
        }

    if user_query.strip() == "":
        return {
            "action": "ask_user",
            "reason": "The user needs to provide a news topic."
        }

    return {
        "action": "tool",
        "reason": "News sources are needed to gather current information."
    }


user_query = "Find recent news about Artificial Intelligence."


decision = create_decision(user_query)

is_valid, validation_message = validate_decision(decision)


print("Behaviour Specification:")
print(BEHAVIOUR_SPEC)

print()
print("Agent Decision:")
print(decision)

print()
print("Validation:")
print(validation_message)
```
